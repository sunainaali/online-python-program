```python
class Employee:
    def __init__(self, name, number, address, salary, post, joining_year, education, experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.joining_year = joining_year
        self.education = education
        self.experience = experience
    def describe_employee(self):
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Joining Year:", self.joining_year)
        print("Education:", self.education)
        print("Experience:", self.experience, "years")
        print("-----------------")
    def update_year_of_experience(self, years):
        self.experience = self.experience + years
        print("New experience of",self.name,"is",self.experience,"years")
    def increase_salary(self, amount):
        self.salary = self.salary + amount
        print("New salary of",self.name,"is",self.salary)
```

```python
employee101 = Employee("Ali khan", 101, "Karachi", 50000, "Manager", 2018, "MBA", 6)
print(employee101.name)
print(employee101.number)
print(employee101.salary)
print(employee101.post)
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(1)
employee101.describe_employee()
```

```python
def make_employee(name, number, address, salary, post, joining_year, education, experience):
    employee = {"name": name, "number": number, "address": address, "salary": salary, "post": post, "joining_year": joining_year, "education": education, "experience": experience}
    return employee
def describe_employee(employee):
    print("Name:", employee["name"])
    print("Number:", employee["number"])
    print("Address:", employee["address"])
    print("Salary:", employee["salary"])
    print("Post:", employee["post"])
    print("Joining Year:", employee["joining_year"])
    print("Education:", employee["education"])
    print("Experience:", employee["experience"], "years")
    print("----------")
def update_year_of_experience(employee, years):
    employee["experience"] = employee["experience"] + years
    print("New experience of", employee["name"], "is", employee["experience"], "years")
def increase_salary(employee, amount):
    employee["salary"] = employee["salary"] + amount
    print("New salary of", employee["name"], "is", employee["salary"])
```

```python
employee101 = make_employee("Ali Khan", 101, "Karachi", 50000, "Manager", 2018, "MBA", 6)
print(employee101["name"])
print(employee101["number"])
print(employee101["salary"])
print(employee101["post"])
describe_employee(employee101)
increase_salary(employee101, 5000)
update_year_of_experience(employee101, 1)
describe_employee(employee101)
```
