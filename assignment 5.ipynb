```python
def count_vowels(text):
    vowels="aeiou"
    count=0  
    for char in text:
        if char in vowels:
            count += 1 
    return f"{text}{count} vowels" 
print(count_vowels("output:"))
```

```python
def display_prices(product,**prices):
    print(f"Product:{product}")
    for store, price in prices.items():
        print(f"{store}:${price}")
display_prices("Laptop", Amazon=800, eBay=750, Walmart=780)
```

```python
def find_max(*numbers):
    highest = max(numbers)
    return highest
print(find_max(10,20,80,100,99,4))
```

```python
def calculate_rectangle(length,width):
    def calculate_area():
        return length*width
    def calculate_perimeter():
        return(length+width)*2
    return calculate_area(), calculate_perimeter()
area,perimeter=calculate_rectangle(5, 3)
print("Area:",area)
print("Perimeter:",perimeter)
```

```python
def hotelbill(night):
    def total():
        room = night * 5000
        tax = room * 0.10
        bill = room + tax  
        return room,tax,bill
    room,tax,total = total()
    
    print("Room Cost:",room)
    print("Tax:",tax)
    print("Total Bill:",total)
hotelbill(3)
```

```python
def login(username,password):
    def details():
        correct_user = "sunaina ali"
        correct_pass = "python"
        if username==correct_user and password==correct_pass:
            return "Access Granted"
        else:
            return "Access Denied"
    status = details()
    return status
username = input("Enter username:")
passkey = input("Enter password:")
result = login(username, passkey)
print(result)
```

```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        return price*quantity
    def apply_discount(total):
        final = total-discount
        if final<0:
            return 0
        else:
            return final
    total = calculate_total()
    final_bill = apply_discount(total)
    return final_bill
bill1 = shopping_bill(200,3) 
print("Final Bill =", bill1, "PKR" " " "(without discount)")
bill2 = shopping_bill(200,3,100) 
print("Final Bill =", bill2, "PKR" " " "(with discount)")
```
